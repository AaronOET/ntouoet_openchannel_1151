# Normal and critical depth of a circle channel

Geometric functions from Table 5.6.1 (Chow, 1959, Table 2.1):
 $d_0$ = pipe diameter (table: $d_o$), $\theta$ = water-surface angle (rad), $y$ = flow depth, $T$ = top width
(table: $B$), $n$ = Manning's roughness, $S_0$ = bed slope, $Q$ = discharge.

| Quantity | Expression |
|---|---|
| Area $A$ | $\tfrac{1}{8}(\theta - \sin\theta)\,d_0^2$ |
| Wetted perimeter $P$ | $\tfrac{1}{2}\theta\,d_0$ |
| Hydraulic radius $R$ | $\tfrac{1}{4}\left(1 - \dfrac{\sin\theta}{\theta}\right) d_0$ |
| Top width $T$ | $\sin(\theta/2)\,d_0 = 2\sqrt{y(d_0 - y)}$ |
| $\dfrac{2}{3R}\dfrac{dR}{dy} + \dfrac{1}{A}\dfrac{dA}{dy}$ | $\dfrac{4(2\sin\theta + 3\theta - 5\theta\cos\theta)}{3d_0\,\theta(\theta - \sin\theta)\sin(\theta/2)}$ |

where $\theta = 2\cos^{-1}\left(1 - \dfrac{2y}{d_0}\right)$ is the angle subtended at the centre by the water surface.

## Method

The functions of the table and the solvers are in `channel_sections.py`; the plot and
the sliders are in `channel_plot.py`.

**Normal depth.** Manning's equation, $Q = \dfrac{1}{n} A R^{2/3} \sqrt{S_0}$, is solved for $y_n$ with
the Newton–Raphson method

$$F(y) = A R^{2/3} - \frac{n\,Q}{\sqrt{S_0}} = 0, \qquad
\frac{dF}{dy} = A R^{2/3} \left(\frac{2}{3R}\frac{dR}{dy} + \frac{1}{A}\frac{dA}{dy}\right),
\qquad y^{k+1} = y^k - \frac{F(y^k)}{dF/dy}$$

where the bracketed term is the last row of Table 5.6.1 (`gfun`).

**Critical depth.** At critical flow the Froude number $Fr = V/\sqrt{g A/T} = 1$, i.e.

$$F(y) = \frac{A^3}{T} - \frac{Q^2}{g} = 0, \qquad
\frac{dF}{dy} = 3A^2 - \frac{A^3}{T^2}\frac{dT}{dy} \quad (\text{using } dA/dy = T)$$

For a circular section $A R^{2/3}$ reaches its maximum at $y \approx 0.938\,d_0$, so a
part-full normal depth exists only for $Q \le Q_{max}$ (`max_discharge`).

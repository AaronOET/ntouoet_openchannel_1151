# Normal and critical depth of a triangle channel

Geometric functions from Table 5.6.1 (Chow, 1959, Table 2.1):
 $s$ = side slope H:1V (table: $z$), $y$ = flow depth, $T$ = top width
(table: $B$), $n$ = Manning's roughness, $S_0$ = bed slope, $Q$ = discharge.

| Quantity | Expression |
|---|---|
| Area $A$ | $s y^2$ |
| Wetted perimeter $P$ | $2y\sqrt{1+s^2}$ |
| Hydraulic radius $R$ | $\dfrac{s y}{2\sqrt{1+s^2}}$ |
| Top width $T$ | $2 s y$ |
| $\dfrac{2}{3R}\dfrac{dR}{dy} + \dfrac{1}{A}\dfrac{dA}{dy}$ | $\dfrac{8}{3y}$ |

## Method

The functions of the table and the solvers are in `channel_sections.py`; the plot and
the sliders are in `channel_plot.py`.

**Normal depth.** Manning's equation, $Q = \dfrac{1}{n} A R^{2/3} \sqrt{S_0}$, is solved for $y_n$ with
the Newton–Raphson method

$$F(y) = A R^{2/3} - \frac{n\,Q}{\sqrt{S_0}} = 0, \qquad
\frac{dF}{dy} = A R^{2/3} \left(\frac{2}{3R}\frac{dR}{dy} + \frac{1}{A}\frac{dA}{dy}\right),
\qquad y^{k+1} = y^k - \frac{F(y^k)}{dF/dy}$$

where the bracketed term is the last row of Table 5.6.1 (`gfun`).

**Critical depth.** At critical flow the Froude number $Fr = V/\sqrt{g A/T} = 1$, i.e.

$$F(y) = \frac{A^3}{T} - \frac{Q^2}{g} = 0, \qquad
\frac{dF}{dy} = 3A^2 - \frac{A^3}{T^2}\frac{dT}{dy} \quad (\text{using } dA/dy = T)$$

## Check of the derivative term against a finite difference
